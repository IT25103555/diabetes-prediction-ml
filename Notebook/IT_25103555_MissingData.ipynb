## 1. Why this technique is needed
The Pima Indians Diabetes dataset has **no NaN values on the surface**, but several
clinically impossible zeros hide missing data: a real patient cannot have
`Glucose = 0`, `BloodPressure = 0`, `SkinThickness = 0`, `Insulin = 0`, or `BMI = 0`.
These zeros are placeholders for unmeasured/unrecorded values, so they must be
treated as missing (NaN) and imputed, otherwise the model learns a false
relationship between "zero" and non-diabetic outcomes.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv"
df = pd.read_csv(url)
print(df.shape)
df.head()
```

```python
df.describe()
```

```python
zero_as_missing_cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

missing_counts = (df[zero_as_missing_cols] == 0).sum()
print("Zeros (= missing) per column:\n", missing_counts)
```

```python
df_missing = df.copy()
df_missing[zero_as_missing_cols] = df_missing[zero_as_missing_cols].replace(0, np.nan)

print(df_missing.isnull().sum())
```

```python
df_imputed = df_missing.copy()
for col in zero_as_missing_cols:
    median_val = df_imputed[col].median()
    df_imputed[col] = df_imputed[col].fillna(median_val)

df_imputed.isnull().sum()
```

## EDA Visualization: Before vs After Imputation (Insulin)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["Insulin"], bins=30, ax=axes[0], color="salmon")
axes[0].set_title("Insulin - Raw (zeros = missing)")
sns.histplot(df_imputed["Insulin"], bins=30, ax=axes[1], color="seagreen")
axes[1].set_title("Insulin - After Median Imputation")
plt.tight_layout()
os.makedirs(os.path.dirname("results/eda_visualizations/missing_data_insulin.png") or ".", exist_ok=True)
plt.savefig("results/eda_visualizations/missing_data_insulin.png", dpi=150)
plt.show()
```
